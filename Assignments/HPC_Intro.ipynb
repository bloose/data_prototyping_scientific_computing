If this connection is successful, then you can proceed to setting up your Python environment in the next step.

**NOTE:** The Oscar Unity environment uses PyModules to let the user customize their work environment and even install software via conda or pip.  Type ```]$ module available``` to see what's available.

### Step 2: Set up the working environment.

* [Do this only once] Once you have loaded the Anaconda module, you need to install the Python libraries that we will use.  This command creates an Anaconda programming environment called `your_env_here`.  You will need to activate this environment every time you log in.
~~~
$ module load conda/latest
$ conda create -n [your_env_here] python=3.9 dask mpi4py scikit-image matplotlib
$ conda activate [your_env_here] # Activate your environment before installing landsatexplore using pip.
~~~


* [Do this every time you log in] Unity uses modules; modules can be loaded with the load command.  You will load them every time you log in.
~~~
$ module load conda/latest
$ source activate [your_env_here]
~~~
At the end of the session when you log out, you can enter  `source deactivate` to end the Anaconda environment.


**RECAP:**

The next time you log in, you won't have to carry out the installation steps.  you will just execute the following command sequence.

```
$ module load conda/latest
$ conda activate [your_env_here]
```
